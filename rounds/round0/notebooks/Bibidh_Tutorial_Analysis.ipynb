```python
import pandas as pd
import os
import pathlib
```

```python
#pathlib.Path().resolve().parent / 'data/prices_round_0_day_-1.csv'
df_day1 = pd.read_csv(pathlib.Path().resolve().parent / 'data/prices_round_0_day_-1.csv', sep = ';')
df_day2 = pd.read_csv(pathlib.Path().resolve().parent / 'data/prices_round_0_day_-2.csv', sep = ';')
df = pd.concat([df_day1,df_day2])
df.head()
```

```python
df_em = df[df['product'] == 'EMERALDS'].copy()
df_tom = df[df['product'] == 'TOMATOES'].copy()
```

```python
df_tom.plot( x = 'timestamp', y ='mid_price')
```

```python
df_tom['mid_price'].diff(periods=1).plot()
```
